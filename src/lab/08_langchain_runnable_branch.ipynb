```python
from langchain_core.runnables import RunnableBranch

branch = RunnableBranch(
    (
        lambda x: "رمز" in x,
        lambda x: "این سؤال مربوط به مدیریت رمز عبور است."
    ),
    (
        lambda x: "مرخصی" in x,
        lambda x: "این سؤال مربوط به مرخصی است."
    ),
    lambda x: "موضوع سؤال مشخص نیست."
)

print(branch.invoke("چطور رمز عبورم را تغییر بدهم؟"))
print(branch.invoke("چطور مرخصی ثبت کنم؟"))
print(branch.invoke("ساعت کاری چیست؟"))
```
