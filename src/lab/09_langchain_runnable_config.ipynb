```python
from langchain_core.runnables import RunnableLambda


def check_user(question, config):
    metadata = config.get("metadata", {})

    user_id = metadata.get("user_id")
    department = metadata.get("department")

    print(f"[AUTH] user={user_id}")
    print(f"[AUTH] department={department}")

    return question


chain = RunnableLambda(check_user)

result = chain.invoke(
    "گزارش اعتباری شعبه را بده",
    config={
        "metadata": {
            "user_id": 125,
            "department": "credit"
        },
        "tags": ["authorization"]
    }
)

print(result)
```

```python
from langchain_core.runnables import RunnableLambda


def generate(question, config):
    model = config["configurable"].get("model", "qwen3:1.7b")

    return f"Question: {question} ==> Model: {model}"


chain = RunnableLambda(generate)

print(
    chain.invoke(
        "RAG چیست؟",
        config={
            "configurable": {
                "model": "qwen3:1.7b"
            }
        }
    )
)
```
