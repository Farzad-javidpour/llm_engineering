```python
from pydantic import BaseModel, Field
from langchain_core.tools import StructuredTool


# -------------------------
# 1. تعریف Schema ورودی
# -------------------------

class CustomerInput(BaseModel):

    customer_id: int = Field(
        description="شماره مشتری"
    )

    include_transactions: bool = Field(
        default=False,
        description="آیا تراکنش‌های مشتری هم برگردانده شود؟"
    )


# -------------------------
# 2. تابع اصلی Tool
# -------------------------

def get_customer(
    customer_id: int,
    include_transactions: bool
) -> dict:

    customer = {
        "customer_id": customer_id,
        "name": "فرزاد احمدی",
        "level": "Gold"
    }

    if include_transactions:
        customer["transactions"] = [
            {"amount": 1000000, "type": "deposit"},
            {"amount": 500000, "type": "withdrawal"}
        ]

    return customer


# -------------------------
# 3. ساخت StructuredTool
# -------------------------

tool = StructuredTool.from_function(
    func=get_customer,
    name="get_customer",
    description="اطلاعات مشتری را بر اساس شماره مشتری برمی‌گرداند.",
    args_schema=CustomerInput
)


# -------------------------
# 4. اجرای Tool
# -------------------------

result = tool.invoke({
    "customer_id": 1001,
    "include_transactions": False
})

print(result)
```
