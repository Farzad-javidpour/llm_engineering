```python
from langchain_core.tools import BaseTool


class GetCustomerTool(BaseTool):

    name: str = "get_customer"

    description: str = (
        "اطلاعات یک مشتری را بر اساس شماره مشتری برمی‌گرداند."
    )

    def _run(self, customer_id: int) -> dict:

        # فعلاً دیتابیس واقعی نداریم
        customers = {
            1001: {
                "name": "فرزاد احمدی",
                "level": "Gold"
            },
            1002: {
                "name": "علی رضایی",
                "level": "Platinum"
            }
        }

        return customers.get(
            customer_id,
            {"error": "Customer not found"}
        )


tool = GetCustomerTool()

result = tool.invoke({
    "customer_id": 1002
})

print(result)
```
