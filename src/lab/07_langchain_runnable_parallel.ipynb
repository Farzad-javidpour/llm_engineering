```python
from langchain_core.runnables import (
    RunnableParallel,
    RunnablePassthrough,
    RunnableLambda
)

def get_context(question):
    return "RAG ترکیبی از بازیابی اطلاعات و تولید پاسخ است."

retriever = RunnableLambda(get_context)

chain = RunnableParallel(
    question=RunnablePassthrough(),
    context=retriever
)

result = chain.invoke("RAG چیست؟")

print(result)
```
