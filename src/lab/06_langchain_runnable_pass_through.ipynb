```python
from langchain_core.runnables import (
    RunnablePassthrough,
    RunnableLambda
)
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
به سؤال زیر با استفاده از Context پاسخ بده.

Question:
{question}

Context:
{context}
""")

def get_context(question):
    return "RAG اطلاعات مرتبط را از منابع بازیابی می‌کند."

context = RunnableLambda(get_context)

chain = {
    "question": RunnablePassthrough(),
    "context": context
} | prompt

result = chain.invoke("RAG چیست؟")

print(result)
```
