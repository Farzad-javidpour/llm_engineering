```python
from pydantic import BaseModel, Field
from langchain.tools import tool


class CustomerInput(BaseModel):

    customer_id: int = Field(
        description="شماره یکتای مشتری"
    )

    include_transactions: bool = Field(
        default=False,
        description="آیا تراکنش‌های مشتری نیز برگردانده شود؟"
    )


@tool(args_schema=CustomerInput)
def get_customer(
    customer_id: int,
    include_transactions: bool = False
) -> dict:
    """اطلاعات مشتری را بر اساس شماره مشتری برمی‌گرداند."""

    return {
        "customer_id": customer_id,
        "include_transactions": include_transactions
    }


print(get_customer.args_schema.model_json_schema())
```
