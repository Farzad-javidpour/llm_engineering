```python
from langchain_core.runnables import RunnableLambda


def create_answer(question, language):
    return f"[{language}] {question}"


runnable = RunnableLambda(create_answer)

# language را ثابت می‌کنیم
persian_runnable = runnable.bind(language="fa")


# حالا فقط question را می‌دهیم
result = persian_runnable.invoke(
    "RAG چیست؟"
)

print(result)
```
