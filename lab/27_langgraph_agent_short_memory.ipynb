```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama
from langgraph.checkpoint.memory import InMemorySaver


# -----------------------------
# Model
# -----------------------------

model = ChatOllama(
    model="qwen3:1.7b",
    temperature=0,
)


# -----------------------------
# Checkpointer
# -----------------------------

checkpointer = InMemorySaver()


# -----------------------------
# Agent
# -----------------------------

agent = create_agent(
    model=model,
    tools=[],
    checkpointer=checkpointer,
)


# -----------------------------
# Thread ID
# -----------------------------

config = {
    "configurable": {
        "thread_id": "conversation-001"
    }
}


# -----------------------------
# پیام اول
# -----------------------------

result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "اسم من علی است."
            }
        ]
    },
    config,
)

print(result["messages"][-1].content)


# -----------------------------
# پیام دوم
# همان Thread
# -----------------------------

result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "اسم من چیست؟"
            }
        ]
    },
    config,
)

print(result["messages"][-1].content)
```
