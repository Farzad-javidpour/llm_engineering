```python
from langchain.tools import tool


@tool
def get_customer(
    customer_id: int,
    include_transactions: bool = False
) -> dict:
    """اطلاعات مشتری را بر اساس شماره مشتری برمی‌گرداند."""

    customers = {
        1001: {
            "name": "فرزاد احمدی",
            "level": "Gold"
        },
        1002: {
            "name": "علی رضایی",
            "level": "Platinum"
        }
    }

    customer = customers.get(customer_id)

    if customer is None:
        return {
            "error": "Customer not found"
        }

    if include_transactions:
        customer["transactions"] = [
            {
                "amount": 1000000,
                "type": "deposit"
            },
            {
                "amount": 500000,
                "type": "withdrawal"
            }
        ]

    return customer


result = get_customer.invoke({
    "customer_id": 1001,
    "include_transactions": True
})

print(result)
```
