```python
from typing import TypedDict

from langchain.agents import create_agent
from langchain.messages import AnyMessage
from langchain_ollama import ChatOllama


# ---------------------------------------------------------
# 1. تعریف State
# ---------------------------------------------------------

class AgentState(TypedDict):
    messages: list[AnyMessage]
    user_id: str
    branch_code: str


# ---------------------------------------------------------
# 2. ساخت Model
# ---------------------------------------------------------

model = ChatOllama(
    model="qwen3:1.7b",
    temperature=0,
)


# ---------------------------------------------------------
# 3. ساخت Agent
# ---------------------------------------------------------

agent = create_agent(
    model=model,
)


# ---------------------------------------------------------
# 4. اجرای Agent
# ---------------------------------------------------------

state = {
    "messages": [
        {
            "role": "user",
            "content": "سلام، وضعیت شعبه من را بررسی کن."
        }
    ],

    "user_id": "U10025",
    "branch_code": "101",
}


result = agent.invoke(state)


# ---------------------------------------------------------
# 5. مشاهده State خروجی
# ---------------------------------------------------------

print("User ID:")
print(result["user_id"])

print()

print("Branch Code:")
print(result["branch_code"])

print()

print("Messages:")
for message in result["messages"]:
    print(type(message).__name__)
    print(message.content)
    print("-" * 50)
```
